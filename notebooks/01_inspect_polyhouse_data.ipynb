# Inspect Polyhouse Sensor Data

Load the raw sensor CSV, inspect its structure, and display the saved Parquet snapshot for comparison.

## Column Definitions

| Column | Definition | Unit |
| --- | --- | --- |
| `timestamp` | Date of the sensor readings and daily yield | Date |
| `temperature_c` | Polyhouse air temperature | Degrees Celsius (°C) |
| `humidity_pct` | Relative humidity inside the polyhouse | Percent (%) |
| `co2_ppm` | Carbon dioxide concentration | Parts per million (ppm) |
| `yield_kg` | Daily mushroom yield (prediction target) | Kilograms (kg) |

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

# Find the project root whether Jupyter starts at the root or in notebooks/.
project_root = next(
    (
        candidate
        for candidate in (Path.cwd(), *Path.cwd().parents)
        if (candidate / "data" / "raw" / "polyhouse_sensors.csv").is_file()
    ),
    None,
)
if project_root is None:
    raise FileNotFoundError(
        "Could not find data/raw/polyhouse_sensors.csv. Run the sample generator first."
    )

raw_path = project_root / "data" / "raw" / "polyhouse_sensors.csv"
snapshot_path = project_root / "data" / "interim" / "01_loaded.parquet"
header = pd.read_csv(raw_path, nrows=0)
date_columns = ["timestamp"] if "timestamp" in header.columns else []
df = pd.read_csv(raw_path, parse_dates=date_columns)
```

## Raw CSV Inspection

```python
print("Shape:", df.shape)
print("\nData types:")
display(df.dtypes.to_frame(name="dtype"))
print("\nDataFrame info:")
df.info()
print("\nFirst five rows:")
display(df.head())
print("\nDataset:")
display(df)
```

## Saved Snapshot

```python
if snapshot_path.is_file():
    snapshot_df = pd.read_parquet(snapshot_path)
    print("Parquet snapshot shape:", snapshot_df.shape)
    display(snapshot_df.head())
else:
    print("Parquet snapshot not found. Run src/ingest.py to create it.")
```
